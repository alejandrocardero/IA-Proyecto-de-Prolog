A continuación, ejecuto un código en Python que usa la librería Pandas. Esto demuestra que sé manipular y analizar datos en este entorno, usando un nivel de conocimiento más avanzado que lo básico. La tabla de datos que muestro simula los datos de nuestro árbol familiar en Prolog.

```python


import pandas as pd

# 1. Crear un diccionario de datos para el "Árbol Familiar"
#    Esto simula la información que podríamos tener en el proyecto Prolog
datos_familia = {
    'Nombre': ['Juan', 'María', 'Carlos', 'Ana', 'Laura'],
    'Relacion': ['Padre', 'Hija', 'Hijo', 'Nieta', 'Nieta'],
    'Edad': [65, 40, 38, 15, 10],
    'Rol_IA': ['Abuelo', 'Padre', 'Padre', 'Hijo', 'Hijo']
}

# 2. Crear una tabla con los datos
df = pd.DataFrame(datos_familia)

print("--- DataFrame Inicial (Tabla de Datos) ---")
print(df)
print("-" * 40)

# 3. Manipulación de Datos (Cálculo simple)
#    Calcular la edad promedio de los "Padres" en esta lista
promedio_edad_padres = df[df['Rol_IA'] == 'Padre']['Edad'].mean()

print(f"\nEdad promedio de los 'Padres' en la lista: {promedio_edad_padres:.1f} años")
```
